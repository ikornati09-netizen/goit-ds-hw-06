```python
import pandas as pd
import numpy as np

df = pd.read_csv("Housing.csv")
print(df.head())
```

```python
X = df[["area", "bathrooms", "bedrooms"]]
y = df["price"]

print(X.head())
print(y.head())
```

```python
X_b = np.c_[np.ones((X.shape[0], 1)), X]

print(X_b[:5])
print(X_b.shape)
```

```python
def hypothesis(X, w):
    return np.dot(X, w)
    
w = np.zeros(4)
predictions = hypothesis(X, w)

print(predictions[:5])
print(predictions.shape)
```

```python
def compute_cost(y, predictions):
    diff = (y - predictions)**2 
    return diff.mean()

cost = compute_cost(y, predictions)
print(cost)
```

```python
X = df[["area", "bathrooms", "bedrooms"]].to_numpy()
bias = np.ones(len(X))
X = np.column_stack((X,bias))
print (X)
print(X.shape)
```

```python
def gradient_descent_step(X, y, w, alpha):
    m = len(y)
    error = hypothesis(X, w) - y
    gradient = 1/m * (X.T @ error)
    w = w - alpha * gradient
    return w  

alpha = 0.01

initial_predictions = hypothesis(X, np.zeros(4))
initial_cost = compute_cost(y, initial_predictions)

print("Initial cost:", initial_cost)

for i in range(1000):
    w=gradient_descent_step(X, y, w, alpha)

    if i % 100 == 0:
        predictions = hypothesis(X, w)
        cost = compute_cost(y, predictions)
        print(i, cost)
        
print("Weights:", w)
print("Cost:", cost)
```

```python
print("Means:", X[:, :3].mean(axis=0))
print("Standard deviations:", X[:, :3].std(axis=0))

X_scaled = X.copy()

mean = X[:, :3].mean(axis=0)
std = X[:, :3].std(axis=0)

X_scaled[:, :3] = (X[:, :3] - mean) / std

print(X_scaled[:5])
print("New means:", X_scaled[:, :3].mean(axis=0))
print("New std:", X_scaled[:, :3].std(axis=0))
```

```python
w = np.zeros(4)
alpha = 0.01

for i in range(1000):
    w = gradient_descent_step(X_scaled, y, w, alpha)

    if i % 100 == 0:
        predictions = hypothesis(X_scaled, w)
        cost = compute_cost(y, predictions)
        print(i, cost)

print("Weights:", w)
print("Cost:", cost)
```

```python
sqrt_cost = np.sqrt(cost)
print(sqrt_cost)
```

```python
w_analytical = np.linalg.pinv(X_scaled) @ y
print("Analytical weights:", w_analytical)
```

```python
sk_predictions = model.predict(X_scaled[:, :3])
sk_cost = compute_cost(y, sk_predictions)
sk_rmse = np.sqrt(sk_cost)

print("Sklearn RMSE:", sk_rmse)
```
